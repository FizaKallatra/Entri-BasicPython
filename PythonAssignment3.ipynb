```python
import random
secret_number = random.randint(1, 10)
print("Welcome to the Number Guessing Game!")
print("I'm thinking of a number between 1 and 10.You have 3 turns to guess the number.")
attempts=3

while(attempts>0):
  a=int(input("Guess the number:"))

  if (a<1 or a>10):                   #Out of Range
    print("Guess is out of range")
    continue
  elif(attempts==1):                  #Out of attempts
    print("You have no more attempts left."
    "The secret number was:",secret_number)
    break
  elif(a==secret_number):             #Correct Guess
    print("Congratulations! You guessed the number correctly.")
    break
  elif (a>secret_number):           #Number is higher than secret no
    print("Too high! Try again.")
  elif(a<secret_number):            #Number is lower than secret no
    print("Too low! Try again.")

  else:
    print("Invalid Input")
  attempts-=1
```
