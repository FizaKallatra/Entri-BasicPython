```python
import numpy as np
```

```python
#1. Create a 1D NumPy Array
```

```python
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
#2. Inspection and Properties:
```

```python
print("Shape:", np.shape(temperatures_w1))
print("Size:", np.size(temperatures_w1))
print("Data Type:",type(temperatures_w1))
print("Data Type of elements:",type(temperatures_w1[0]))
```

```python
#3. Array Operations:
```

```python
temp_f=temperatures_w1*(9/5)+32
print("Temperature in Fahrenheit:",temp_f)
```

```python
print("Mean temperature of the week:",np.mean(temp_f))
print("Maximum temperature of the week:",np.max(temp_f))
print("Minimum temperature of the week:",np.min(temp_f))
```

```python
#4. Array Slicing and Indexing:
```

```python
print("Temperature for 1st 3 days of the week:",temperatures_w1[:3])
print("Temperature for last 2 days of the week:",temperatures_w1[-2:])
print("Temperature of middle 3 days:",temperatures_w1[2:5])
```

```python
#5. Create a 2D Array:
```

```python
temperatures=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
```

```python
print(temperatures)
```

```python
print("Shape:",np.shape(temperatures))
print("Size:",np.size(temperatures))
print("Data Type:",np.dtype(temperatures[0,1]))
```

```python
print("The temperature for week 1:",temperatures[0,:])
print("The temperature for the 1st weekend:",temperatures[0:1,5:7])
print("The temperature for week 2:",temperatures[1,:])
print("The temperature for the 2nd weekend:",temperatures[1:2,5:7])
```

```python
"""Pandas Series:
1. Creating Pandas Series:
"""
```

```python
import pandas as pd
```

```python
marks=pd.Series([95,92,89,85,80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
print("First Rank Student:",marks.loc['Rank1'])
```

```python
print("First 3 Rank Student:\n",marks.loc['Rank1':'Rank3'])
```

```python
print("Third Rank Student:",marks.iloc[2])
```

```python
high_marks = marks[marks > 90]
print("Ranks with marks > 90:")
print(high_marks)
```

```python
#3. Manipulating Series
```

```python
marks['Rank1']=100
print(marks)
```

```python
marks=marks.drop('Rank5')
print(marks)
```

```python
cgpa=marks/10
print(cgpa)
```

```python
#Pandas DataFrame
```

```python
transactions={"TransactionID":[101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
              "ProductCategory":['Electronics', 'Clothing', 'Electronics', 'Furniture','Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
              "Region":['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West','South', 'North'],
              "Amount":[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]}

df=pd.DataFrame(transactions)
print(df)
```

```python
#2. Data Exploration
```

```python
print("Shape:",df.shape)
print("Size:",df.size)
print("ColumnName:",df.columns)
print("\nData Type:\n",df.dtypes)
print("\nHead:\n",df.head())
print("\nTail:\n",df.tail())
```

```python
print(df[["ProductCategory","Amount"]])
```

```python
print("Last 3 columns:\n",df.iloc[:, -3:])
```

```python
print(df[(df["Region"]=="North") & (df["Amount"]>200)])
```

```python
print(df['ProductCategory'].value_counts())
```

```python
print("Unique values in Region:",df["Region"].unique())
```

```python
print(df.groupby("Region")["Amount"].mean())
```

```python
#3. Manipulating the DataFrame
```

```python
df.loc[df["TransactionID"] == 102, "Amount"] = 165
print(df)
```

```python
df["Discount"]=df["Amount"] * 0.10
print(df)
```

```python
df = df[df["TransactionID"] != 109]
print(df)
```

```python
df=df.drop("Discount",axis=1)
print(df)
```
